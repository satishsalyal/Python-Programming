## 0. Introduction — Functions vs Methods

Python gives us ready-made tools to work with lists. They come in two flavours:

| | **Function** | **Method** |
|---|---|---|
| Called as | `function(list)` | `list.method()` |
| Examples | `len(lst)`, `sorted(lst)`, `max(lst)` | `lst.append(5)`, `lst.sort()` |
| Belongs to | Python in general | the list object itself |

**Important idea — in-place vs. new result**

* Many list **methods change the original list** (they *mutate* it) and usually return `None`: `append, extend, insert, remove, clear, reverse, sort`.
* Some **return a value**: `pop()`, `index()`, `count()`.
* Some **functions give a new result and leave the list unchanged**: `sorted()`, `list()`.

You can see every available list method with `dir()`:

```python
print([m for m in dir(list) if not m.startswith("__")])
```

## 1. `len()` — number of elements

**Syntax:** `len(list)`  
**Returns:** an integer — the total number of elements (items at the top level).

```python
marks = [85, 90, 78, 92]
print(len(marks))               # 4
print(len([]))                  # 0 -> empty list
print(len(["apple"]))           # 1
```

### Points to remember
* A nested list counts as **one** element.
* The last valid index is always `len(lst) - 1`.

```python
nested = [1, [2, 3, 4], 5]
print(len(nested))              # 3, not 5 - the inner list counts as one

lst = [10, 20, 30, 40]
print("Last element:", lst[len(lst) - 1])

# Typical use: loop over every index
for i in range(len(lst)):
    print(i, lst[i])
```

## 2. `list()` — create / convert to a list

**Syntax:** `list()` or `list(iterable)`  
Creates a **new list**. With no argument it makes an empty list; with an iterable (string, tuple, range, set, dictionary …) it converts it to a list.

```python
print(list())                        # []  empty list
print(list("HELLO"))                 # string -> characters
print(list((1, 2, 3)))               # tuple  -> list
print(list(range(5)))                # range  -> [0, 1, 2, 3, 4]
print(list(range(2, 11, 2)))         # [2, 4, 6, 8, 10]
print(list({10, 20, 30}))            # set -> list (order not guaranteed)
```

### Dictionary → list
By default you get the **keys**. Use `.values()` or `.items()` for the others.

```python
d = {"a": 1, "b": 2, "c": 3}
print(list(d))             # keys
print(list(d.values()))    # values
print(list(d.items()))     # (key, value) pairs
```

### `list()` makes a *new* list — also a way to copy

```python
a = [1, 2, 3]
b = list(a)          # shallow copy
b.append(4)
print("a =", a)      # unchanged
print("b =", b)

# Common error: a single number is not iterable
try:
    list(123)
except TypeError as e:
    print("TypeError:", e)

print(list(str(123)))      # ['1', '2', '3']  - convert to string first
```

### Creating a list of numbers from user input

```python
text = "10 20 30 40"                       # imagine this came from input()
numbers = list(map(int, text.split()))
print(numbers)
```

## 3. `append()` — add one element at the end

**Syntax:** `list.append(item)`  
* Adds **exactly one** item at the **end**.  
* Changes the original list; returns `None`.  
* If the item is itself a list, it is added **as a single element** (nested).

```python
fruits = ["apple", "banana"]
fruits.append("mango")
print(fruits)

fruits.append("orange")
print(fruits)
```

### Appending different types

```python
items = []
items.append(10)
items.append("hello")
items.append(3.5)
items.append(True)
items.append([1, 2])         # whole list added as ONE element
print(items)
print(len(items))            # 5
```

### `append()` returns `None` ⚠️

```python
a = [1, 2, 3]
b = a.append(4)
print(a)      # [1, 2, 3, 4]
print(b)      # None   <- do not write  a = a.append(4)
```

### Building a list with a loop (very common)

```python
squares = []
for n in range(1, 6):
    squares.append(n * n)
print(squares)

# only even numbers
evens = []
for n in range(1, 11):
    if n % 2 == 0:
        evens.append(n)
print(evens)
```

## 4. `extend()` — add many elements at the end

**Syntax:** `list.extend(iterable)`  
* Takes an **iterable** (list, tuple, string, set, range …) and adds **each of its items** one by one at the end.  
* Changes the original list; returns `None`.

```python
a = [1, 2, 3]
a.extend([4, 5, 6])
print(a)

a.extend((7, 8))            # tuple
print(a)

a.extend(range(9, 12))      # range
print(a)
```

### ⚠️ Extending with a string adds every *character*

```python
letters = ["a", "b"]
letters.extend("xyz")
print(letters)              # ['a', 'b', 'x', 'y', 'z']

names = ["Asha"]
names.extend(["Ravi"])       # to add one whole string, wrap it in a list
print(names)

try:
    [1, 2].extend(5)         # int is not iterable
except TypeError as e:
    print("TypeError:", e)
```

### `extend()` is similar to `+=`

```python
x = [1, 2]
x += [3, 4]       # same effect as x.extend([3, 4])
print(x)
```

## 5. Difference between `append()` and `extend()`

| Feature | `append()` | `extend()` |
|---|---|---|
| Argument | any **one object** | an **iterable** |
| Adds | the object **as a single element** | **each item** of the iterable |
| Effect on length | always `+1` | `+ len(iterable)` |
| `lst.append([4, 5])` | `[1, 2, 3, [4, 5]]` | — |
| `lst.extend([4, 5])` | — | `[1, 2, 3, 4, 5]` |
| Argument can be an int? | ✅ yes | ❌ `TypeError` |

```python
base = [1, 2, 3]

a = base.copy()
a.append([4, 5])
print("append :", a, " length =", len(a))

e = base.copy()
e.extend([4, 5])
print("extend :", e, " length =", len(e))

# with a string
s1 = ["a"]; s1.append("bcd")
s2 = ["a"]; s2.extend("bcd")
print("append('bcd') :", s1)
print("extend('bcd') :", s2)
```

## 6. `insert()` — add an element at a given position

**Syntax:** `list.insert(index, item)`  
* Inserts `item` **before** the element currently at `index`.  
* Existing elements from that position are shifted to the right.  
* Modifies the original list; returns `None`.

```python
nums = [10, 20, 30, 40]

nums.insert(1, 15)       # before index 1
print(nums)

nums.insert(0, 5)        # at the beginning
print(nums)

nums.insert(len(nums), 50)   # at the end (same as append)
print(nums)
```

### Special cases
* Index **greater than length** → item is added at the **end** (no error).
* **Negative** index counts from the end; `insert(-1, x)` places `x` **before** the last element.
* Only **one** item is inserted — a list is inserted as a single element.

```python
a = [1, 2, 3]
a.insert(100, 99)
print(a)                 # [1, 2, 3, 99]

b = [1, 2, 3]
b.insert(-1, 99)
print(b)                 # [1, 2, 99, 3]

c = [1, 2, 3]
c.insert(1, [8, 9])
print(c)                 # [1, [8, 9], 2, 3]

d = [1, 2, 3]
d.insert(-100, 0)        # very small negative index -> start of list
print(d)
```

### Keeping a list sorted while inserting

```python
sorted_list = [10, 20, 40, 50]
new = 30

pos = 0
while pos < len(sorted_list) and sorted_list[pos] < new:
    pos += 1
sorted_list.insert(pos, new)
print(sorted_list)
```

### `append()` vs `insert()`
`append` → always the end. `insert` → any position (slower for large lists because items must shift).

## 7. `index()` — position of an element

**Syntax:** `list.index(item [, start [, end]])`  
* Returns the **index of the first occurrence** of `item`.  
* Raises **`ValueError`** if the item is not present.  
* Optional `start` and `end` limit the search range.

```python
colors = ["red", "green", "blue", "green", "yellow"]

print(colors.index("blue"))          # 2
print(colors.index("green"))         # 1  -> first occurrence only
print(colors.index("green", 2))      # 3  -> search from index 2 onwards
print(colors.index("green", 2, 5))   # 3  -> search in [2, 5)
```

### ValueError when the item is missing

```python
try:
    colors.index("pink")
except ValueError as e:
    print("ValueError:", e)

# Safe approach: check with 'in' first
item = "pink"
if item in colors:
    print("Found at", colors.index(item))
else:
    print(item, "not in list")
```

### Finding *all* positions of a value
`index()` gives only the first one, so use a loop (or `enumerate`).

```python
positions = []
for i, c in enumerate(colors):
    if c == "green":
        positions.append(i)
print(positions)                     # [1, 3]

# one-liner
print([i for i, c in enumerate(colors) if c == "green"])
```

## 8. `pop()` — remove and return an element by index

**Syntax:** `list.pop()` or `list.pop(index)`  
* Without an argument it removes the **last** element.  
* With an index it removes the element at that position.  
* **Returns** the removed element (so you can store it).  
* `IndexError` if the list is empty or the index is invalid.

```python
nums = [10, 20, 30, 40, 50]

last = nums.pop()
print("Removed:", last, "| list:", nums)

second = nums.pop(1)
print("Removed:", second, "| list:", nums)

print(nums.pop(-1), nums)       # negative index works too
```

### Errors

```python
try:
    [].pop()
except IndexError as e:
    print("IndexError:", e)

try:
    [1, 2, 3].pop(10)
except IndexError as e:
    print("IndexError:", e)
```

### Using `pop()` as a stack (LIFO) 

```python
stack = []
stack.append("page1")
stack.append("page2")
stack.append("page3")
print("Stack:", stack)

while stack:
    print("Popped:", stack.pop())
```

## 9. `remove()` — remove an element by value

**Syntax:** `list.remove(value)`  
* Removes the **first occurrence** of `value`.  
* Returns `None`.  
* Raises **`ValueError`** if `value` is not found.

| `pop(i)` | `remove(x)` |
|---|---|
| deletes by **position** | deletes by **value** |
| returns the removed item | returns `None` |
| `IndexError` if bad index | `ValueError` if not found |

```python
fruits = ["apple", "banana", "mango", "banana", "orange"]

fruits.remove("banana")
print(fruits)                  # only the FIRST banana removed

fruits.remove("banana")
print(fruits)                  # now the second one
```

```python
try:
    fruits.remove("grapes")
except ValueError as e:
    print("ValueError:", e)

# Safe removal
if "grapes" in fruits:
    fruits.remove("grapes")
else:
    print("grapes not present, nothing removed")
```

### Removing *all* occurrences of a value

```python
nums = [1, 2, 3, 2, 4, 2, 5]

# Method 1: while loop with membership test
a = nums.copy()
while 2 in a:
    a.remove(2)
print(a)

# Method 2: build a new list (preferred)
b = [n for n in nums if n != 2]
print(b)
```

### ⚠️ Do not remove while looping over the same list

```python
data = [2, 2, 3, 2, 4]
for x in data:             # modifying list while iterating -> skips items
    if x == 2:
        data.remove(x)
print("Wrong:", data)       # [2, 3, 4] - not all 2s were removed

data = [2, 2, 3, 2, 4]
for x in data[:]:          # loop over a COPY
    if x == 2:
        data.remove(x)
print("Right:", data)
```

## 10. `clear()` — remove all elements

**Syntax:** `list.clear()`  
* Deletes **all items**, leaving an **empty list** `[]`.  
* The list object still **exists** (same name, same memory object).  
* Returns `None`.

```python
nums = [1, 2, 3, 4]
print("id before:", id(nums))

nums.clear()
print(nums, "| length =", len(nums))
print("id after :", id(nums))     # same object, just emptied

nums.append(10)                    # still usable
print(nums)
```

## 11. `del` — delete elements or the whole list

`del` is a **Python statement** (not a method), so no parentheses and no `.` notation.

| Usage | Effect |
|---|---|
| `del lst[i]` | removes the element at index `i` |
| `del lst[a:b]` | removes a slice |
| `del lst[::k]` | removes every k-th element (extended slice) |
| `del lst` | deletes the **variable itself** |

```python
nums = [10, 20, 30, 40, 50, 60, 70, 80]

del nums[0]              # single element
print(nums)

del nums[1:3]            # slice (index 1 and 2)
print(nums)

del nums[-1]             # last element
print(nums)

del nums[::2]            # every second element
print(nums)
```

### Deleting the entire list

```python
temp = [1, 2, 3]
del temp                 # the name 'temp' no longer exists

try:
    print(temp)
except NameError as e:
    print("NameError:", e)
```

### Errors

```python
try:
    del [1, 2, 3][10]
except IndexError as e:
    print("IndexError:", e)
```

## 12. Difference between `del` and `clear()`

| Feature | `del lst` | `lst.clear()` |
|---|---|---|
| Type | statement | method |
| Result | the **list variable is removed** | list becomes **empty** `[]` |
| Using the name afterwards | ❌ `NameError` | ✅ works normally |
| Other references to the list | unaffected (only the name is deleted) | **all** references see the empty list |
| Can remove only part? | ✅ `del lst[1:3]` | ❌ always everything |

```python
a = [1, 2, 3]
b = a                   # b refers to the same list

a.clear()
print("after clear -> a:", a, "| b:", b)    # both empty (same object)

c = [1, 2, 3]
d = c
del c                   # removes only the NAME c
print("after del   -> d:", d)               # d still holds [1, 2, 3]
try:
    print(c)
except NameError:
    print("c no longer exists")
```

### Quick comparison of removal tools

```python
def fresh(): return [10, 20, 30, 40, 50]

x = fresh(); x.pop();        print("pop()        ->", x)
x = fresh(); x.pop(1);       print("pop(1)       ->", x)
x = fresh(); x.remove(30);   print("remove(30)   ->", x)
x = fresh(); del x[1:3];     print("del x[1:3]   ->", x)
x = fresh(); x.clear();      print("clear()      ->", x)
```

## 13. `count()` — how many times a value occurs

**Syntax:** `list.count(value)`  
Returns an integer; returns `0` if the value is absent (no error).

```python
nums = [1, 2, 3, 2, 4, 2, 5]
print(nums.count(2))        # 3
print(nums.count(5))        # 1
print(nums.count(99))       # 0

words = ["to", "be", "or", "not", "to", "be"]
print(words.count("to"))
print(words.count("To"))    # 0  -> case-sensitive
```

### Counting in nested lists & finding duplicates

```python
mixed = [[1, 2], [3], [1, 2], 5, 5]
print(mixed.count([1, 2]))   # 2
print(mixed.count(5))        # 2

# Items that appear more than once
data = ["a", "b", "a", "c", "b", "a"]
for item in set(data):
    print(item, "->", data.count(item))
```

## 14. `reverse()` — reverse the list in place

**Syntax:** `list.reverse()`  
* Reverses the order of elements **in the same list**.  
* Returns `None`.  
* For a **new** reversed list use `lst[::-1]` or `reversed(lst)`.

```python
nums = [1, 2, 3, 4, 5]
nums.reverse()
print(nums)

x = [1, 2, 3]
print(x.reverse())          # None
print(x)

# Without changing the original
orig = [1, 2, 3, 4]
print(orig[::-1], list(reversed(orig)), orig)
```

## 15. `sort()` — sort the list in place

**Syntax:** `list.sort(key=None, reverse=False)`  
* Sorts **ascending** by default; `reverse=True` gives descending.  
* `key` is a function applied to each element to decide the order.  
* Changes the original list; returns `None`.  
* All elements must be **comparable** (can't mix numbers and strings).

```python
nums = [45, 12, 78, 3, 56]

nums.sort()
print("ascending :", nums)

nums.sort(reverse=True)
print("descending:", nums)
```

### Sorting strings
Uppercase letters sort **before** lowercase (based on character codes). Use `key=str.lower` for a case-insensitive sort.

```python
names = ["Zoya", "amit", "Bhavna", "chirag"]

names.sort()
print(names)

names.sort(key=str.lower)
print(names)

words = ["banana", "kiwi", "apple", "fig"]
words.sort(key=len)                  # by length
print(words)
```

### ⚠️ Mixed types cannot be sorted; `sort()` returns `None`

```python
try:
    [3, "a", 1].sort()
except TypeError as e:
    print("TypeError:", e)

x = [3, 1, 2]
y = x.sort()
print(y)        # None   (don't write x = x.sort())
print(x)
```

## 16. `sorted()` — get a new sorted list

**Syntax:** `sorted(iterable, key=None, reverse=False)`  
* Works on **any iterable** (list, tuple, string, set, dict keys …).  
* Always returns a **new list**; the original is **unchanged**.

```python
nums = [45, 12, 78, 3, 56]

new = sorted(nums)
print("sorted   :", new)
print("original :", nums)                 # unchanged
print("descending:", sorted(nums, reverse=True))

print(sorted("python"))                   # string -> sorted list of characters
print(sorted((9, 4, 7)))                  # tuple  -> list
print(sorted({"b": 2, "a": 1, "c": 3}))   # dict keys
print(sorted(["Zoya", "amit", "Bhavna"], key=str.lower))
```

## 17. Difference between `sort()` and `sorted()`

| Feature | `list.sort()` | `sorted(iterable)` |
|---|---|---|
| Kind | list **method** | built-in **function** |
| Changes original? | ✅ **Yes** (in place) | ❌ No |
| Returns | `None` | a **new list** |
| Works on | lists only | any iterable |
| Memory | no extra list | creates a new list |
| Use when | you don't need the original order | you want to keep the original |

```python
a = [3, 1, 2]
r1 = a.sort()
print("sort()  -> returned:", r1, "| list:", a)

b = [3, 1, 2]
r2 = sorted(b)
print("sorted()-> returned:", r2, "| list:", b)

# sort() cannot be used on a tuple, sorted() can
t = (3, 1, 2)
print(sorted(t))
print(hasattr(t, "sort"))      # False
```

## 18. `max()` and `min()` — largest and smallest

**Syntax:** `max(list)`, `min(list)` (also work on tuples, strings, etc.)  
* Numbers → numerical comparison. Strings → alphabetical (by character code).  
* Empty list → `ValueError` (unless you give `default=`).  
* Optional `key=` function.

```python
nums = [45, 12, 78, 3, 56]
print("Max:", max(nums))
print("Min:", min(nums))

words = ["banana", "apple", "cherry"]
print(max(words), min(words))                 # alphabetical

print(max(words, key=len))                    # longest word
print(min(words, key=len))                    # shortest word

print(max(3, 9, 5))                           # also works with separate values
```

### Errors and the `default` argument

```python
try:
    max([])
except ValueError as e:
    print("ValueError:", e)

print(max([], default=0))            # safe for empty lists

try:
    max([1, "a"])
except TypeError as e:
    print("TypeError:", e)
```

### Without `max()`/`min()` — logic using a loop

```python
nums = [45, 12, 78, 3, 56]
big = small = nums[0]
for n in nums:
    if n > big:   big = n
    if n < small: small = n
print(big, small)
```

## 19. `sum()` — total of the elements

**Syntax:** `sum(iterable, start=0)`  
* Adds all **numeric** elements; `start` is added to the total (default `0`).  
* Empty list → `0`.  
* Strings can't be summed → `TypeError`.

```python
marks = [85, 90, 78, 92]
print("Total  :", sum(marks))
print("Average:", sum(marks) / len(marks))
print("With start=100:", sum(marks, 100))
print(sum([]))                        # 0
print(sum([1.5, 2.5, 3.0]))           # 7.0
```

```python
try:
    print(sum(["a", "b"]))
except TypeError as e:
    print("TypeError:", e)

# Convert strings to numbers first
raw = ["10", "20", "30"]
print(sum(int(x) for x in raw))

# Sum of nested lists (flatten idea)
print(sum([[1, 2], [3, 4]], []))      # [1, 2, 3, 4]  - list concatenation trick
```

## 20. Putting it all together — Mini Programs

### 20.1 Student marks manager

```python
marks = []

# add marks
marks.append(72)
marks.extend([85, 90, 65])
marks.insert(0, 88)
print("Marks          :", marks)

# analysis
print("Total students :", len(marks))
print("Highest        :", max(marks))
print("Lowest         :", min(marks))
print("Average        :", sum(marks) / len(marks))
print("Ranking (high->low):", sorted(marks, reverse=True))

# remove the lowest score
marks.remove(min(marks))
print("After removing lowest:", marks)

# who scored 85 and where
if 85 in marks:
    print("85 found at index", marks.index(85))
```

### 20.2 To-do list

```python
todo = ["study", "exercise"]
todo.append("cook")
todo.insert(1, "email")
print(todo)

done = todo.pop(0)
print("Finished:", done)
print("Pending :", todo)

todo.clear()
print("All cleared:", todo)
```

### 20.3 Remove duplicates and sort

```python
data = [5, 3, 5, 1, 3, 9, 1]

unique = []
for x in data:
    if data.count(x) >= 1 and x not in unique:
        unique.append(x)

unique.sort()
print(unique)

# shorter way
print(sorted(set(data)))
```

### 20.4 Accept numbers from the user and analyse them

```python
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

print("Count   :", len(numbers))
print("Sum     :", sum(numbers))
print("Max/Min :", max(numbers), "/", min(numbers))
print("Sorted  :", sorted(numbers))
print("Reversed:", numbers[::-1])
```

## 21. Summary — Cheat Sheet

| Name | Syntax | Changes list? | Returns |
|---|---|---|---|
| `len` | `len(lst)` | no | number of elements |
| `list` | `list(iterable)` | creates new | new list |
| `append` | `lst.append(x)` | ✅ | `None` |
| `extend` | `lst.extend(iter)` | ✅ | `None` |
| `insert` | `lst.insert(i, x)` | ✅ | `None` |
| `index` | `lst.index(x)` | no | first index (`ValueError` if absent) |
| `pop` | `lst.pop([i])` | ✅ | removed element |
| `remove` | `lst.remove(x)` | ✅ | `None` (`ValueError` if absent) |
| `clear` | `lst.clear()` | ✅ | `None` |
| `del` | `del lst[i]` / `del lst[a:b]` / `del lst` | ✅ | — (statement) |
| `count` | `lst.count(x)` | no | occurrences (0 if absent) |
| `reverse` | `lst.reverse()` | ✅ | `None` |
| `sort` | `lst.sort(key, reverse)` | ✅ | `None` |
| `sorted` | `sorted(iter, key, reverse)` | no | new sorted list |
| `max` / `min` | `max(lst)` / `min(lst)` | no | largest / smallest |
| `sum` | `sum(lst, start)` | no | total |

**Key differences to remember**

* `append(x)` adds **one object**; `extend(iter)` adds **each item**.
* `pop(i)` removes by **index and returns**; `remove(x)` removes by **value**.
* `del lst` removes the **variable**; `lst.clear()` **empties** it.
* `sort()` is **in place & returns None**; `sorted()` returns a **new list**.
