# 🐍  `if` / `if-else` Control Statements in Python

This notebook covers conditional logic:
chained conditions, nested conditions, `elif` ladders, logical operators,
ternary expressions, and membership checks.
Run each cell and read the explanation above it.

## 1️⃣ Age Category Classifier (if / elif / else ladder)

**Concept:** A multi-branch `elif` ladder. Python checks each condition **top to bottom** and executes the **first** branch that is `True`, skipping everything after it.
- **Intermediate trick:** Order matters! Always check the most specific conditions first.

```python
age = 25

if age < 13:
    print("Child")
elif age < 20:
    print("Teenager")
elif age < 65:
    print("Adult")
else:
    print("Senior")
```

## 2️⃣ Grade Calculator (chained comparisons)

**Concept:** Chained comparisons let you write `a <= x <= b` instead of two separate comparisons joined by `and`.
- Cleaner syntax, and both sides are checked **only once**.

```python
score = 87

if 90 <= score <= 100:
    grade = "A"
elif 80 <= score < 90:
    grade = "B"
elif 70 <= score < 80:
    grade = "C"
elif score >= 0:
    grade = "Fail"
else:
    grade = "Invalid score"

print("Grade:", grade)
```

## 3️⃣ Leap Year Checker (compound logical conditions)

**Concept:** Use `and` / `or` to combine conditions. A year is a leap year if it is divisible by 4 **and** (not divisible by 100 **or** divisible by 400).
- Parentheses control the order of evaluation when `and` and `or` are mixed.

```python
year = 2024

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year")
else:
    print(f"{year} is NOT a leap year")
```

## 4️⃣ Login System (nested if)

**Concept:** An `if` inside another `if` = **nested conditional**. The inner block only runs if the outer condition is `True`.
- Use nesting when a condition depends on a previous condition being true.

```python
username = "admin"
password = "secret123"
stored_user = "admin"
stored_pass = "secret123"

if username == stored_user:
    if password == stored_pass:
        print("✅ Login successful")
    else:
        print("❌ Wrong password")
else:
    print("❌ Username not found")
```

## 5️⃣ Number Sign & Even-Odd Check (nested with logical operators)

**Concept:** Combine nesting with `and` to test multiple properties of a value.
- Flow: first check if it's zero, then check sign, then check parity.

```python
num = -14

if num != 0:
    sign = "positive" if num > 0 else "negative"
    if num % 2 == 0:
        print(f"{num} is an even {sign} number")
    else:
        print(f"{num} is an odd {sign} number")
else:
    print("The number is zero")
```

## 6️⃣ Discount Calculator (nested if with arithmetic)

**Concept:** Real-world pricing logic — apply discounts based on **membership status** and **purchase amount**.
- VIP members get a bigger discount, but only above a minimum purchase.

```python
is_member = True
purchase = 1200

if is_member:
    if purchase >= 1000:
        discount = 0.20
    elif purchase >= 500:
        discount = 0.10
    else:
        discount = 0.05
else:
    if purchase >= 1000:
        discount = 0.10
    else:
        discount = 0.0

final_price = purchase * (1 - discount)
print(f"Discount applied: {discount*100:.0f}%")
print(f"Final price: ${final_price:.2f}")
```

## 7️⃣ Triangle Validity & Type Check

**Concept:** A triangle is valid only if the sum of **any** two sides is greater than the third.
- After validating, an `elif` ladder classifies it as Equilateral, Isosceles, or Scalene.

```python
a, b, c = 5, 5, 8

if a + b > c and a + c > b and b + c > a:
    print("Valid triangle")
    if a == b == c:
        print("Type: Equilateral")
    elif a == b or b == c or a == c:
        print("Type: Isosceles")
    else:
        print("Type: Scalene")
else:
    print("Not a valid triangle")
```

## 8️⃣ ATM Withdrawal (multiple guard conditions)

**Concept:** Use an `if` with multiple `and`/`or` conditions as **guards** before allowing an action.
- Each failed guard produces a different, meaningful error message.

```python
balance = 5000
withdraw = 4500
daily_limit = 4000

if withdraw <= 0:
    print("❌ Amount must be positive")
elif withdraw > balance:
    print("❌ Insufficient balance")
elif withdraw > daily_limit:
    print("❌ Exceeds daily limit of", daily_limit)
else:
    balance -= withdraw
    print(f"✅ Withdrawn: ${withdraw} | Remaining balance: ${balance}")
```

## 9️⃣ Voting Eligibility (ternary + logical operators)

**Concept:** The **ternary (conditional) expression**: `value_if_true if condition else value_if_false`.
- It is a one-line `if-else`. Also demonstrates how to build complex conditions with `not` and `or`.

```python
age = 17
has_id = True

# Ternary expression
status = "Can vote" if age >= 18 and has_id else "Cannot vote"
print(status)

# Nested ternary for a 3-way decision
category = "Minor" if age < 18 else ("Adult" if age < 65 else "Senior")
print("Category:", category)
```

## 🔟 Membership & Access Control (in operator + if-else)

**Concept:** Use the `in` operator to check membership in a list/tuple — much cleaner than multiple `or` comparisons.
- Combine with `else` to create a simple role-based access system.

```python
role = "editor"
blocked_users = ["hacker1", "spam_bot"]

if role in blocked_users:
    print("🚫 Access denied")
elif role in ["admin", "editor"]:
    print("✅ Full content access")
elif role in ["viewer", "guest"]:
    print("👀 Read-only access")
else:
    print("❓ Unknown role — default read-only access")
```

## 🎁 Bonus: `match-case` (Python 3.10+)

**Concept:** For many exact-value branches, `match-case` is often cleaner than a long `elif` ladder.

```python
command = "start"

match command:
    case "start":
        print("Starting the engine...")
    case "stop":
        print("Stopping the engine...")
    case "restart":
        print("Restarting the engine...")
    case _:
        print("Unknown command")
```

---
### ✅ Key Takeaways
1. **`elif` ladders** → first `True` condition wins; order matters.
2. **Chained comparisons** → `90 <= score <= 100`.
3. **Nested `if`** → when a condition depends on another.
4. **Logical operators** → `and`, `or`, `not` to combine conditions.
5. **Ternary** → `x if cond else y` for one-liners.
6. **`in` operator** → clean membership checks.
7. **`match-case`** → modern alternative to long `elif` chains (Python 3.10+).
